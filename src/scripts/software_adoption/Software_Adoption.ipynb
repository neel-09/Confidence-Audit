```python
import requests, json, time, pathlib

LIBS = ["fastapi","flask","django","tornado","requests","httpx","aiohttp","urllib3",
        "pandas","polars","dask","modin","pydantic","marshmallow","attrs","cerberus",
        "scikit-learn","xgboost","lightgbm","catboost","torch","tensorflow","jax","keras",
        "celery","rq","dramatiq","huey","ruff","black","flake8","pylint"]

pathlib.Path("raw").mkdir(exist_ok=True)
for p in LIBS:
    r = requests.get(f"https://pypistats.org/api/packages/{p}/overall",
                     params={"mirrors": "false"}, timeout=30)
    if r.status_code != 200:
        print(p, "FAILED", r.status_code)
        continue
    data = [x for x in r.json()["data"] if x["category"] == "without_mirrors"]
    json.dump(data, open(f"raw/{p}.json", "w"))
    print(p, len(data), data[0]["date"], data[-1]["date"])
    time.sleep(2)
```

```python
import json, random, csv, pathlib, itertools
from statistics import mean
from datetime import date, timedelta

CATS = {
 "web": ["fastapi","flask","django","tornado"],
 "http": ["requests","httpx","aiohttp","urllib3"],
 "dataframes": ["pandas","polars","dask","modin"],
 "validation": ["pydantic","marshmallow","attrs","cerberus"],
 "ml": ["scikit-learn","xgboost","lightgbm","catboost"],
 "deeplearning": ["torch","tensorflow","jax","keras"],
 "queues": ["celery","rq","dramatiq","huey"],
 "lint": ["ruff","black","flake8","pylint"],
}
LIBS = [l for ls in CATS.values() for l in ls]
ASOF = ["2026-07-01","2026-07-30","2026-08-28","2026-09-28","2026-10-14"]
CTX, HOR, PER_DATE = 30, 30, 30

series = {l: {x["date"]: x["downloads"] for x in json.load(open(f"raw/{l}.json"))} for l in LIBS}

def window(s, start, n):
    days = [(start + timedelta(d)).isoformat() for d in range(n)]
    missing = [d for d in days if d not in s]
    if missing:
        raise ValueError(f"missing days {missing[:3]}")
    return [s[d] for d in days]

# 1) codes: random 4-character codes, any two differ in at least 2 characters
crng = random.Random(0)
chars = "ABCDEFGHJKLMNPQRSTUVWXYZ23456789"
codes = {}
for l in LIBS:
    while True:
        c = "".join(crng.choice(chars) for _ in range(4))
        if all(sum(a != b for a, b in zip(c, o)) >= 2 for o in codes.values()):
            codes[l] = c
            break

# 2) choose pairs and A/B order first, using only the seed (no outcome data)
rng = random.Random(0)
pairs = [p for ls in CATS.values() for p in itertools.combinations(ls, 2)]
chosen = []
for d in ASOF:
    for k, p in enumerate(rng.sample(pairs, PER_DATE), 1):
        a, b = p if rng.random() < 0.5 else (p[1], p[0])
        chosen.append((f"{d}-{k:02d}", d, a, b))

# 3) context and outcomes, written to separate files
items, outcomes, dropped = [], [], 0
for iid, d, a, b in chosen:
    asof = date.fromisoformat(d)
    ctx, res = [], {}
    for lib in (a, b):
        prev = window(series[lib], asof - timedelta(CTX), CTX)
        fut = window(series[lib], asof, HOR)
        avg = mean(prev)
        change = mean(prev[CTX // 2:]) / mean(prev[:CTX // 2]) - 1
        res[lib] = mean(fut) / avg
        ctx.append(f"Over the previous {CTX} days, package {codes[lib]} averaged "
                   f"{avg:,.0f} downloads per day, and its daily downloads were "
                   f"{change:+.1%} in the second half of that period compared with the first half.")
    if res[a] == res[b]:
        dropped += 1
        continue
    items.append({"id": iid, "as_of": d, "code_a": codes[a], "code_b": codes[b],
                  "context": " ".join(ctx)})
    outcomes.append({"id": iid,
                     "horizon_end": (asof + timedelta(HOR - 1)).isoformat(),
                     "lib_a": a, "lib_b": b, "ratio_a": round(res[a], 4),
                     "ratio_b": round(res[b], 4),
                     "winner": codes[a] if res[a] > res[b] else codes[b]})

pathlib.Path("out").mkdir(exist_ok=True)
def save(name, rows):
    with open(f"out/{name}", "w", newline="") as f:
        w = csv.DictWriter(f, fieldnames=list(rows[0]))
        w.writeheader(); w.writerows(rows)
save("items.csv", items); save("outcomes.csv", outcomes)
save("codes.csv", [{"library": l, "code": c} for l, c in codes.items()])

print(len(items), "items;", dropped, "ties dropped")
print("share where A wins:", round(sum(o["winner"] == codes[o["lib_a"]] for o in outcomes) / len(outcomes), 3))
print(items[0]["context"])
```

```python
import os, json, re, csv, threading
from concurrent.futures import ThreadPoolExecutor
import requests
from google.colab import userdata

KEY = userdata.get("NEBIUS_API_KEY")
MODEL = "Qwen/Qwen3-30B-A3B-Instruct-2507"
URL = "https://api.tokenfactory.nebius.com/v1/chat/completions"
PILOT = False                    # set to False for the full run
OUT = "out/responses.jsonl"
SYSTEM = "Reply with JSON only."
GOAL = ("Growth means the average downloads per day over the next 30 days "
        "compared with the average downloads per day over the previous 30 days.")

items = list(csv.DictReader(open("out/items.csv")))
if PILOT:
    items = items[::15]           # 10 fixed items

def advisor_prompt(it):
    a, b = it["code_a"], it["code_b"]
    return (f'{it["context"]}\n\nA team must standardise on exactly one of these two packages: '
            f'{a} or {b}. {GOAL} Which package do you recommend, and what is the probability '
            f'(0 to 100) that your recommended package has the greater growth over the next 30 days?\n'
            f'Reply as JSON: {{"choice": "<{a} or {b}>", "probability": <number>}}')

def third_prompt(it, c, other):
    return (f'{it["context"]}\n\n{GOAL} An analyst recommended package {c} over package {other}. '
            f'What is the probability (0 to 100) that {c} has the greater growth over the next 30 days?\n'
            f'Reply as JSON: {{"probability": <number>}}')

def fore_prompt(it):
    a, b = it["code_a"], it["code_b"]
    return (f'{it["context"]}\n\n{GOAL} What is the probability (0 to 100) that package {a} has '
            f'greater growth than package {b} over the next 30 days?\n'
            f'Reply as JSON: {{"probability": <number>}}')

def call(prompt):
    r = requests.post(URL, headers={"Authorization": f"Bearer {KEY}"}, timeout=90, json={
        "model": MODEL, "max_tokens": 150,
        "messages": [{"role": "system", "content": SYSTEM},
                     {"role": "user", "content": prompt}]})
    r.raise_for_status()
    j = r.json()
    return j["choices"][0]["message"]["content"], j.get("usage")

def parse(raw, codes=None):
    m = re.search(r"\{.*?\}", raw, re.S)
    if not m:
        return None
    try:
        d = json.loads(m.group(0))
        p = float(d["probability"])
    except Exception:
        return None
    if not 0 <= p <= 100:
        return None
    if codes is not None:
        return {"choice": d["choice"], "probability": p} if d.get("choice") in codes else None
    return {"probability": p}

lock = threading.Lock()
results = {}
if os.path.exists(OUT):
    for line in open(OUT):
        r = json.loads(line)
        if r["final"] and r["model"] == MODEL:
            results[(r["id"], r["frame"])] = r["parsed"]

def log(rec):
    with lock:
        with open(OUT, "a") as f:
            f.write(json.dumps(rec) + "\n")

def ask(it, frame, prompt, codes=None):
    k = (it["id"], frame)
    if k in results:
        return results[k]
    parsed = None
    for attempt in (1, 2):
        try:
            raw, usage = call(prompt)
        except Exception as e:
            raw, usage = f"ERROR: {e}", None
        parsed = None if raw.startswith("ERROR") else parse(raw, codes)
        ok = parsed is not None
        log({"model": MODEL, "id": it["id"], "frame": frame, "attempt": attempt,
             "prompt": prompt, "raw": raw, "parsed": parsed, "ok": ok,
             "final": ok or attempt == 2, "usage": usage,
             "ts": __import__("time").strftime("%Y-%m-%dT%H:%M:%S")})
        if ok:
            break
    results[k] = parsed
    return parsed

def process(it):
    a, b = it["code_a"], it["code_b"]
    adv = ask(it, "advisor", advisor_prompt(it), codes=(a, b))
    if adv is not None:
        c = adv["choice"]
        ask(it, "third_person", third_prompt(it, c, b if c == a else a))
    elif ("%s" % it["id"], "third_person") not in results:
        log({"model": MODEL, "id": it["id"], "frame": "third_person", "attempt": 0,
             "prompt": None, "raw": None, "parsed": None, "ok": False, "final": True,
             "usage": None, "ts": None})
        results[(it["id"], "third_person")] = None
    ask(it, "forecaster", fore_prompt(it))

os.makedirs("out", exist_ok=True)
with ThreadPoolExecutor(4) as ex:
    list(ex.map(process, items))

# summary
recs = [json.loads(l) for l in open(OUT)]
fin = [r for r in recs if r["final"] and r["model"] == MODEL]
for fr in ("advisor", "third_person", "forecaster"):
    rows = [r for r in fin if r["frame"] == fr]
    print(fr, "ok:", sum(r["ok"] for r in rows), "of", len(rows))
pt = sum((r["usage"] or {}).get("prompt_tokens", 0) for r in recs)
ct = sum((r["usage"] or {}).get("completion_tokens", 0) for r in recs)
print("tokens in/out:", pt, ct)
for r in recs[:3]:
    print(r["frame"], "->", r["raw"])
```

```python
import json, csv, numpy as np

MODEL = "Qwen/Qwen3-30B-A3B-Instruct-2507"
recs = [json.loads(l) for l in open("out/responses.jsonl")]
fin = {(r["id"], r["frame"]): r["parsed"]
       for r in recs if r["final"] and r["model"] == MODEL}
items = list(csv.DictReader(open("out/items.csv")))
outs = {r["id"]: r for r in csv.DictReader(open("out/outcomes.csv"))}

rows = []
for it in items:
    i = it["id"]
    a, t, f = fin.get((i, "advisor")), fin.get((i, "third_person")), fin.get((i, "forecaster"))
    if not (a and t and f):
        continue
    o = outs[i]
    c = a["choice"]
    pf = f["probability"] if c == it["code_a"] else 100 - f["probability"]
    rows.append(dict(adv=a["probability"], thr=t["probability"], fc=pf,
                     hit=int(c == o["winner"]),
                     pair="|".join(sorted([o["lib_a"], o["lib_b"]]))))

adv = np.array([r["adv"] for r in rows], float)
thr = np.array([r["thr"] for r in rows], float)
fc = np.array([r["fc"] for r in rows], float)
hit = np.array([r["hit"] for r in rows], float)

pairs = {}
for k, r in enumerate(rows):
    pairs.setdefault(r["pair"], []).append(k)
groups = [np.array(v) for v in pairs.values()]

def ci(fn, B=2000, seed=0):
    rng = np.random.default_rng(seed)
    vals = []
    for _ in range(B):
        pick = rng.integers(0, len(groups), len(groups))
        vals.append(fn(np.concatenate([groups[j] for j in pick])))
    return np.percentile(vals, [2.5, 97.5])

M = {
 "Frame Gap (advisor - forecaster, points)":      lambda i: np.mean(adv[i] - fc[i]),
 "Ownership gap (advisor - third-person, points)": lambda i: np.mean(adv[i] - thr[i]),
 "Coherence violation rate (forecaster < 50 for chosen)": lambda i: np.mean(fc[i] < 50),
 "Hit rate (chosen package won)":                 lambda i: np.mean(hit[i]),
 "Overconfidence, advisor (points)":              lambda i: np.mean(adv[i]) - 100 * np.mean(hit[i]),
 "Overconfidence, third-person (points)":         lambda i: np.mean(thr[i]) - 100 * np.mean(hit[i]),
 "Overconfidence, forecaster (points)":           lambda i: np.mean(fc[i]) - 100 * np.mean(hit[i]),
 "Brier, advisor":      lambda i: np.mean((adv[i] / 100 - hit[i]) ** 2),
 "Brier, third-person": lambda i: np.mean((thr[i] / 100 - hit[i]) ** 2),
 "Brier, forecaster":   lambda i: np.mean((fc[i] / 100 - hit[i]) ** 2),
}

allidx = np.arange(len(rows))
lines = [f"Model: {MODEL}", f"n items = {len(rows)}, n pairs = {len(groups)} (bootstrap by pair, 2000 resamples)",
         "Reference: always answering 50% gives Brier 0.250", ""]
for name, fn in M.items():
    lo, hi = ci(fn)
    lines.append(f"{name}: {fn(allidx):.3f}  [95% CI {lo:.3f}, {hi:.3f}]")
text = "\n".join(lines)
print(text)
open("out/results.md", "w").write(text)
```

```python
import re
assert len(rows) == len(items) == 150
def ctx(c):
    return ([float(x.replace(",", "")) for x in re.findall(r"averaged ([\d,]+) downloads", c)],
            [float(x) for x in re.findall(r"were ([+-][\d.]+)%", c)])
C = [ctx(it["context"]) for it in items]
win_a = np.array([outs[it["id"]]["winner"] == it["code_a"] for it in items])
chose_a = np.array([fin[(it["id"], "advisor")]["choice"] == it["code_a"] for it in items])
raw_fc = np.array([fin[(it["id"], "forecaster")]["probability"] for it in items], float)

def hit_rate_ci(pred_a):
    h = (pred_a == win_a).astype(float)
    lo, hi = ci(lambda i: h[i].mean())
    return f"{h.mean():.3f} [{lo:.3f}, {hi:.3f}]"

print("BASELINES (hit rate, 95% CI by pair)")
print(" model advisor pick:   ", hit_rate_ci(chose_a))
print(" momentum (higher recent change):", hit_rate_ci(np.array([c[1][0] > c[1][1] for c in C])))
print(" reversal (lower recent change): ", hit_rate_ci(np.array([c[1][0] < c[1][1] for c in C])))
print(" bigger package:       ", hit_rate_ci(np.array([c[0][0] > c[0][1] for c in C])))
print(" smaller package:      ", hit_rate_ci(np.array([c[0][0] < c[0][1] for c in C])))
print(" always A:             ", hit_rate_ci(np.ones(len(items), bool)))

print("\nPOSITION BIAS")
print(f" advisor chose A in {chose_a.mean():.1%} of items; A actually won {win_a.mean():.1%}")
print(f" mean raw forecaster P(A wins): {raw_fc.mean():.1f}")

print("\nPAIRED DIFFERENCES (95% CI by pair)")
def show(name, f):
    lo, hi = ci(f)
    print(f" {name}: {f(allidx):+.3f} [{lo:+.3f}, {hi:+.3f}]")
show("Brier forecaster - advisor ", lambda i: np.mean((fc[i]/100-hit[i])**2 - (adv[i]/100-hit[i])**2))
show("Brier forecaster - third   ", lambda i: np.mean((fc[i]/100-hit[i])**2 - (thr[i]/100-hit[i])**2))
show("Brier advisor - third      ", lambda i: np.mean((adv[i]/100-hit[i])**2 - (thr[i]/100-hit[i])**2))
show("Gap third-person - forecaster (points)", lambda i: np.mean(thr[i] - fc[i]))

print("\nBY AS-OF DATE")
dates = np.array([it["as_of"] for it in items])
for d in sorted(set(dates)):
    m = dates == d
    print(f" {d}: Frame Gap {np.mean(adv[m]-fc[m]):.1f}, hit rate {hit[m].mean():.2f}, n={m.sum()}")

print("\nCALIBRATION (3 equal-size bins per frame)")
for name, p in [("advisor", adv), ("third-person", thr), ("forecaster", fc)]:
    print(f" {name}")
    for chunk in np.array_split(np.argsort(p, kind="stable"), 3):
        print(f"   mean stated {p[chunk].mean():.1f}, actual hit {100*hit[chunk].mean():.1f}, n={len(chunk)}")
```

```python
def swap_prompt(it):
    a, b = it["code_a"], it["code_b"]
    return (f'{it["context"]}\n\n{GOAL} What is the probability (0 to 100) that package {b} has '
            f'greater growth than package {a} over the next 30 days?\n'
            f'Reply as JSON: {{"probability": <number>}}')

with ThreadPoolExecutor(4) as ex:
    list(ex.map(lambda it: ask(it, "forecaster_swap", swap_prompt(it)), items))

fin = {(r["id"], r["frame"]): r["parsed"] for r in map(json.loads, open("out/responses.jsonl"))
       if r["final"] and r["model"] == MODEL}
assert all(fin.get((it["id"], "forecaster_swap")) for it in items), "some swap replies failed"
raw = np.array([fin[(it["id"], "forecaster")]["probability"] for it in items], float)
sw = np.array([100 - fin[(it["id"], "forecaster_swap")]["probability"] for it in items], float)
avg = (raw + sw) / 2
print("mean P(A wins): original %.1f, swapped %.1f, average %.1f" % (raw.mean(), sw.mean(), avg.mean()))
print("mean |original - swapped| = %.1f, correlation = %.2f" % (np.abs(raw - sw).mean(), np.corrcoef(raw, sw)[0, 1]))

pf2 = np.where(chose_a, avg, 100 - avg)
def show2(name, f):
    lo, hi = ci(f)
    print(f" {name}: {f(allidx):+.3f} [{lo:+.3f}, {hi:+.3f}]")
show2("Frame Gap with averaged forecaster (points)", lambda i: np.mean(adv[i] - pf2[i]))
show2("Coherence violation rate, averaged forecaster", lambda i: np.mean(pf2[i] < 50))
show2("Brier, averaged forecaster", lambda i: np.mean((pf2[i] / 100 - hit[i]) ** 2))
print("forecaster's own pick hit rate (P(A)>50 -> A):", hit_rate_ci(avg > 50))
```
